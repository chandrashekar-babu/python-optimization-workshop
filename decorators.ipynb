## Decorators

```python
def to_upper(fn):
    def wrapper():
        return fn().upper()
    return wrapper

def testfn():
    return "Hello world"



print(testfn)
testfn = to_upper(testfn)
print(testfn)

testfn()
```

```python
def sanitize_number(fn):
    def wrapper(x):
        try:
            x = int(x)
        except:
            x = 0
        return fn(x)
    return wrapper

@sanitize_number
def square(x):
    return x*x

square(3.2)
```

```python
def log_calls(fn):
    from time import ctime
    def wrap(*args, **kwargs):
        with open("call_logs.log", "a") as log:
            ret = fn(*args, **kwargs)
            print(f"{ctime()}: function {fn.__qualname__} invoked with {args=}, {kwargs=}, returned {ret}.", file=log)
        return ret
    return wrap


@log_calls
def square(x):
    return x*x

@log_calls
def greet(user):
    print("Hello", user)


square(2)
greet("John")
square(5.6)
square(4)
greet("Sam")
```

```python
def decorate(fn):
    print("decorate invoked on", fn)
    def wrap():
        print("wrap invoked for", fn)
        fn()
    return wrap

@decorate
def greet():
    print("greet invoked")

greet()
greet()
greet()


```

```python
def to_upper(fn):
    def wrap():
        return fn().upper()
    wrap.__qualname__ = f"@to_upper.{fn.__qualname__}"
    return wrap

def to_strong(fn):
    def wrap():
        return "<strong>" + fn() + "</strong>"
    wrap.__qualname__ = f"@to_strong.{fn.__qualname__}"
    return wrap


@to_strong
@to_upper
def greet():
    return "Hello, world"


greet
```

```python
# Parameterized decorators

def style(s):
    def decorate(fn):
        if s == "upper":
            def wrap():
                return fn().upper()
        elif s == "strong":
            def wrap():
                return "<strong>" + fn() + "</strong>"
        else:
            wrap = fn

        return wrap
    return decorate


@style("upper")
@style("strong")
def greet():
    return "Hello, world"

greet()
```

```python
class Accumulate:
    def __init__(self, v=0):
        self.v = v

    def __call__(self, n=0):
        self.v += n
        return self.v

a = Accumulate(0)
a(10)
a(20)
a(30)
a(40)
a(50)
a()
```

##### Class-based decorators

```python
def to_upper(fn):
    def wrapper():
        return fn().upper()
    return wrapper

class ToUpper:
    def __init__(self, fn):
        self.fn = fn

    def __repr__(self):
        return f"<@Toupper: {self.fn}>"

    def __call__(self):
        return self.fn().upper()

@ToUpper  # testfn = ToUpper(testfn)
def testfn():
    return "Hello world"


testfn()
```

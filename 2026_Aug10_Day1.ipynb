#### Exercise: Word frequency counter

Count the number of occurrence of each word in a given string input

```python
s = "this is a test is a test string with test number of test words"

def count_words(s):
    freq = {}
    for word in s.split():
        freq[word] = freq.get(word, 0) + 1

        #if word not in freq:
        #    freq[word] = 1
        #else:
        #    freq[word] += 1

    for w, c in freq.items():
        print(f"'{w}' occurs {c} times")


count_words(s)
# "'this' occurs 1 times"
# "'is' occurs 2 times"
# "'a' occurs 2 times"
# "'test' occurs 4 times"
# ...
```

```python
s = "this is a test is a test string with test number of test words"
from collections import defaultdict

def count_words(s):
    freq = defaultdict(int)
    for word in s.split():
        freq[word] += 1

        #if word not in freq:
        #    freq[word] = 1
        #else:
        #    freq[word] += 1

    for w, c in freq.items():
        print(f"'{w}' occurs {c} times")


count_words(s)
# "'this' occurs 1 times"
# "'is' occurs 2 times"
# "'a' occurs 2 times"
# "'test' occurs 4 times"
# ...
```

```python
s = "this is a test is a test string with test number of test words"

def count_words(s):
    from collections import Counter
    freq = Counter(s.split())

    for w, c in freq.items():
        print(f"'{w}' occurs {c} times")


count_words(s)
# "'this' occurs 1 times"
# "'is' occurs 2 times"
# "'a' occurs 2 times"
# "'test' occurs 4 times"
# ...
```

```python
data = [5, "hello", False, 4+5j, 4.3, 5, "world", "test", 3.14, -6, 7.89, True]

def organize(data):
    from collections import defaultdict
    result = defaultdict(list)
    for item in data:
        key = type(item).__name__
        result[key].append(item)
    return result

def organize(data):
    result = {}
    for item in data:
        key = type(item).__name__
        result.setdefault(key, []).append(item)
    return result



organize(data)
# OUT: {
#   "int": [5, 5, -6], 
#   "str": ["hello", "world", "test"], 
#   "float": [3.14, 7.89], 
#   "complex: [4+5j]",
#   "bool": [False, True]}
```

#### Multitasking using Generators

```python
from time import sleep

def foo():
    for i in range(10):
        print("In foo: counting", i)
        sleep(1)

def bar():
    for i in range(10):
        print("In bar: counting", i)
        sleep(1)

foo()
bar()
    
```

```python
from time import sleep

def foo():
    for i in range(10):
        print("In foo: counting", i)
        yield
        sleep(1)

def bar():
    for i in range(10):
        print("In bar: counting", i)
        yield
        sleep(1)

t1 = foo()
t2 = bar()


for _ in zip(t1, t2):
    pass
```
